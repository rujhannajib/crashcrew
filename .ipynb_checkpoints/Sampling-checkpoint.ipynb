```python
import pandas as pd

filepath = 'US_Accidents_March23.csv' # Replace with your file path
chunk_size = 100000 # Number of rows to read at a time
sample_fraction = 0.01 # Sample 1% of each chunk

sampled_data_frames = []
for chunk in pd.read_csv(filepath, chunksize=chunk_size):
    sampled_chunk = chunk.sample(frac=sample_fraction)
    sampled_data_frames.append(sampled_chunk)

final_sampled_df = pd.concat(sampled_data_frames)

print(f"Loaded a sample of size: {len(final_sampled_df)} rows")
```

```python
final_sampled_df.to_csv("sampled_accident_data.csv", index=False)
```
